```python
from pyBrainAnalyzIR.vis import NIRSviewIR

import pandas as pd

import pyBrainAnalyzIR
import pyBrainAnalyzIR.testing

import pyBrainAnalyzIR.dataclasses.dataset as dataset
from pyBrainAnalyzIR.vis.demographics_manager import demographics_manager

```

Let's start by just creating a random dataset to play with

```python
dset=dataset.DataSet()
data1,_=pyBrainAnalyzIR.testing.simData.Data(snr=5)
data2,_=pyBrainAnalyzIR.testing.simData.Data(snr=5)
data3,_=pyBrainAnalyzIR.testing.simData.Data(snr=5)
data4,_=pyBrainAnalyzIR.testing.simData.Data(snr=5)
data5,_=pyBrainAnalyzIR.testing.simData.Data(snr=5)
data6,_=pyBrainAnalyzIR.testing.simData.Data(snr=5)

# Add the data to the dataset
dset.import_data(data1)
dset.import_data(data2)
dset.import_data(data3)
dset.import_data(data4)
dset.import_data(data5)
dset.import_data(data6)
```

```python
# Currently, the demographics are stored in the meta_data attribute of the data objects. You can access them like this:

data1.meta_data['subject'] = 'subj_1'
data2.meta_data['subject'] = 'subj_2'
data3.meta_data['subject'] = 'subj_3'
data4.meta_data['subject'] = 'subj_4'
data5.meta_data['subject'] = 'subj_5'
data6.meta_data['subject'] = 'subj_6'
```

```python
# Because Python passes by references, editing the meta_data of the data objects will also update the values in the dataset (or vice versa)

demo = dset.get_demographics()
print(demo)
```

```python
demo=pd.DataFrame({'gender':['M','M','F','F','M','F'],
                   'age':[1.,3.,5.,6.,1.,3.]})

# the method add_demographics_by_index will add the demographics to the dataset by matching the index of the demographics dataframe to the index of the dataset. The index of the demographics dataframe should match the index of the dataset. 
dset.add_demographics_by_index(demo)

demo = dset.get_demographics()
print(demo)
```

```python
demo=pd.DataFrame({'subject':['subj_6','subj_5','subj_4','subj_3','subj_2','subj_1'],
                   'count':[1,2,3,4,5,6]})

# You can also add demographics by matching a variable in the demographics dataframe to a variable in the dataset. In this case, we will match the 'subject' variable in the demographics dataframe to the 'subject' variable in the dataset. The values of the 'subject' variable in the demographics dataframe should match the values of the 'subject' variable in the dataset.
dset.add_demographics_by_match_variable(demo, matchvariable='subject')

demo = dset.get_demographics()
print(demo)
```

```python
demo=pd.DataFrame({'subject':['subj_6','subj_5','subj_4'],
                   'someBool':[True,False,True]})

# This also works if you have missing values in the demographics dataframe. In this case, we will match the 'subject' variable in the demographics dataframe to the 'subject' variable in the dataset. The values of the 'subject' variable in the demographics dataframe should match the values of the 'subject' variable in the dataset. If there are missing values in the demographics dataframe, they will be filled with NaN in the dataset.  Note the allow_missing argument is set to FALSE by default, which will raise an error if there are missing values in the demographics dataframe. If you set allow_missing to TRUE, the missing values will be filled with NaN in the dataset.

dset.add_demographics_by_match_variable(demo, matchvariable='subject',allow_missing=True)

demo = dset.get_demographics()
print(demo)
```
